# Data Cleaning Report

### CGPA Cleaning

- Valid CGPA range: 0–10.
- 7 records contained invalid CGPA values greater than 10.
- No values were below 0.
- Invalid values (11 and 12) were converted to NaN because the correct CGPA could not be determined.
- The original student records were retained.
- Missing CGPA values will be handled separately during the missing-value treatment phase.

```python
import pandas as pd
import numpy as np
df = pd.read_csv('student_placement_messy.csv')
```

## Categorical Cleaning

#### Branch Column Cleaning

```python
df['Branch'].value_counts(dropna=False)
```

```python
df['Branch'] = df['Branch'].str.strip()
```

```python
df['Branch'] = df['Branch'].str.upper()
```

```python
df['Branch'].unique()
```

```python
branch_map = {
    'CIVIL':'Civil', 
    'CSE':'CSE', 
    'ECE':'ECE', 
    'IT':'IT', 
    'MECHANICAL':'Mechanical'
}

df['Branch'] = df['Branch'].map(branch_map)
```

```python
df['Branch'].value_counts(dropna=False)
```

#### Internship Column Cleaning

```python
df['Internship'].value_counts(dropna=False)
```

```python
df['Internship'] = df['Internship'].str.strip()
```

```python
df['Internship'] = df['Internship'].str.capitalize()
```

```python
df['Internship'].value_counts(dropna=False)
```

#### Gender Column Cleaning

```python
df['Gender'].value_counts(dropna=False)
```

```python
df['Gender'] = df['Gender'].str.strip()
```

```python
df['Gender'] = df['Gender'].str.capitalize()
```

```python
df['Gender'].value_counts(dropna = False)
```

#### Placement_Status Column Checking

```python
df['Placement_Status'].value_counts(dropna=False)
```

```python
df['Placement_Status'] = df['Placement_Status'].str.strip()
```

```python
df['Placement_Status'] = df['Placement_Status'].str.capitalize()
```

```python
df['Placement_Status'].value_counts(dropna = False)
```

#### Final Check

```python
final_check = ['Gender','Internship','Branch','Placement_Status']
for col in final_check:
    print(df[col].value_counts(dropna=False))
```

#### Attendance Column Cleaning

```python
df['Attendance'].describe()
```

```python
df['Attendance'] = pd.to_numeric(df['Attendance'],errors='coerce')
```
