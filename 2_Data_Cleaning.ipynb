# Data Cleaning

## Categorical Cleaning

Standardized:
- Branch
- Internship
- Placement_Status
- Gender

Operations:
- Removed leading/trailing spaces
- Standardized capitalization
- Standardized category labels
- Missing categorical values were represented as `Unknown`

## Numeric Cleaning

Validated:
- CGPA: 0–10
- Attendance: 0–100
- 10th Percentage: 0–100
- 12th Percentage: 0–100

Invalid values were identified and converted to missing values where the correct original value could not be determined.

Remaining unknown numeric values were retained as `NaN`.

## Date Cleaning

- Converted `Application_Date` to datetime format.
- Invalid/unavailable dates were retained as `NaT`.

## Duplicate Handling

- Duplicate records were identified and removed.
- Duplicate count was verified after removal.

## Missing Value Policy

Missing values were not blindly replaced.

- Categorical unknowns → `Unknown`
- Numeric unknowns → `NaN`
- Date unknowns → `NaT`

This preserves uncertainty instead of inventing values.

```python
import pandas as pd
import numpy as np
df = pd.read_csv('student_placement_messy.csv')
```

## Categorical Cleaning

#### Branch Column Cleaning

```python
df['Branch'].value_counts(dropna=False)
```

```python
df['Branch'] = df['Branch'].str.strip()
```

```python
df['Branch'] = df['Branch'].str.upper()
```

```python
df['Branch'].unique()
```

```python
branch_map = {
    'CIVIL':'Civil', 
    'CSE':'CSE', 
    'ECE':'ECE', 
    'IT':'IT', 
    'MECHANICAL':'Mechanical'
}

df['Branch'] = df['Branch'].map(branch_map)
```

```python
df['Branch'].value_counts(dropna=False)
```
