### Placement Overview

The dataset contains 1820 students with known placement status.

- Placed: 1311 students (72.03 %)
- Not placed: 493 students (27.08 %)

This provides the baseline placement outcome for further analysis.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('Student_clean_data.csv')
```

## Placement Status

```python
df['Placement_Status'].value_counts()
```

```python
df['Placement_Status'].value_counts(normalize = True)*100
```

#### Actual Placement Rate

```python
known_placement = df[df["Placement_Status"].isin(["Placed", "Not placed"])]
```

```python
placement_rate = ((known_placement['Placement_Status']=='Placed').mean()*100)
```

```python
f"Placement Rate: {placement_rate:.2f}%"
```

## Branch Status

```python
df['Branch'].value_counts()
```

```python
branch_wise_placement_count = pd.crosstab(df["Branch"],df["Placement_Status"])
```

```python
branch_wise_placement_rate = (known_placement.groupby('Branch')['Placement_Status'].apply(lambda x:(x=='Placed').mean()*100).round(2))
```

```python
branch_wise_placement_rate
```
