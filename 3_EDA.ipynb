- Not placed: 493 students (27.08 %)

This provides the baseline placement outcome for further analysis.

## Main Analytics

#### Placement Analytics

```python
df['Placement_Status'].value_counts()
```

```python
df['Placement_Status'].value_counts(normalize=True)*100
```

```python
placement_rate = df['Placement_Status'].value_counts(normalize = True).mul(100).round(2)
```

#### Branch Analytics

```python
branch_count = df['Branch'].value_counts()
```

```python
branch_count
```

```python
branch_placement = pd.crosstab(df['Branch'],df['Placement_Status'])
```

```python
branch_placement
```

```python
branch_placement_rate = df[df["Placement_Status"].isin(["Placed", "Not placed"])].groupby("Branch")['Placement_Status'].apply(lambda x: (x == "Placed").mean() * 100).round(2)
```

```python
branch_placement_rate
```

```python
branch_summary = pd.DataFrame({"Total Students": df["Branch"].value_counts(),"Placement Rate (%)": branch_placement_rate})
branch_summary
```
